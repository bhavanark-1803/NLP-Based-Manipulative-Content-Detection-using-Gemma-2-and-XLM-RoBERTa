> This notebook downloads the data provided by UNLP Shared Task and combines the data for testing set into one file.

```python
import requests
import pandas as pd
```

```python
url_train = "https://raw.githubusercontent.com/unlp-workshop/unlp-2025-shared-task/main/data/techniques_classification/train.parquet"
url_test = "https://raw.githubusercontent.com/unlp-workshop/unlp-2025-shared-task/main/data/techniques_classification/test.csv"
url_solution_classif = "https://raw.githubusercontent.com/unlp-workshop/unlp-2025-shared-task/main/data/techniques_classification/solution.csv"
url_solution_span = "https://raw.githubusercontent.com/unlp-workshop/unlp-2025-shared-task/refs/heads/main/data/span_detection/solution.csv"

local_filename_train = "../data/train.parquet"
local_filename_test = "../data/test.csv"
local_filename_solution_classif = "../data/solution_classif.csv"
local_filename_solution_span = "../data/solution_span.csv"
```

```python
def download_file(url, local_filename):
    response = requests.get(url)
    if response.status_code == 200:
        with open(local_filename, 'wb') as f:
            f.write(response.content)
        print(f"File downloaded successfully as {local_filename}")
    else:
        print(f"Failed to download file. Status code: {response.status_code}")
```

```python
download_file(url_train, local_filename_train)
download_file(url_test, local_filename_test)
download_file(url_solution_classif, local_filename_solution_classif)
download_file(url_solution_span, local_filename_solution_span)
```

```python
train = pd.read_parquet(local_filename_train)
test = pd.read_csv(local_filename_test)
solution_classif = pd.read_csv(local_filename_solution_classif)
solution_span = pd.read_csv(local_filename_solution_span)

display(train.head())
display(test.head())
display(solution_classif.head())
display(solution_span.head())
```

```python
solution_classif['techniques'] = solution_classif.apply(lambda row: [col for col in solution_classif.columns if col != 'id' and col != 'Usage' and row[col] != 0], axis=1)
```

```python
test = test.merge(solution_classif[['id', 'techniques']], on='id', how='left')
display(test.head())
```

```python
test = test.merge(solution_span[['id', 'trigger_words']], on='id', how='left')
display(test.head())
```

```python
test.to_csv('../data/test.csv', index=False)
```
